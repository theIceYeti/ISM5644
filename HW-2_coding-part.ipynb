#### 3. Write a function that takes a list as an argument and returns a list that only contains the items that occurred exactly once in the original list.

```python
def singleList(items):
    result = []
    for i in items:
        count = 0
        for f in items:
            if f == i:
                count += 1
        if count == 1:
            result.append(i)
    return result
print(singleList([1, 2, 2, 3, 4, 4, 5, 5, 6])) 
```

#### 4. Suppose that you are the manager of a manufacturing firm who is trying to determine the optimal price of a product based on demand. The relationship between price and demand is as follows: 
Demand = 800 – 10 x Price
#### Each product costs 50 USD to manufacture. The profit of the company is calculated as: 
Profit = (Price – Cost per item) x Demand
#### Write a function that, given a single price input, will first calculate its demand and then return the profit amount. Then, search a range of prices (using the function above) to identify the optimal price that will provide you the maximum profit. Notice that the range is limited, because the demand goes negative if price exceeds a certain threshold. Print out this price and the maximum profit amount.


```python
def calculateProfit(price):
    cost = 50
    demand = 800 - 10 * price
    profit = (price - cost) * demand
    return profit
bestPrice = 0
maxProfit = 0
for price in range(1, 80):        
    profit = calculateProfit(price)
    if profit > maxProfit:
        maxProfit = profit
        bestPrice = price
print("Optimal price:", bestPrice)
print("Maximum profit:", maxProfit)
```

#### 6. You are given a list variable containing the daily closing stock prices for a specific company over a 5-day period:<br>
stock_prices = [150.25, 153.40, 149.80, 152.10, 155.00]<br>

Write a program that repeatedly prompts the user to enter an integer index to retrieve the stock price for a specific day. Because users can make mistakes, wrap your input conversion and list access within a try-except block to build a robust program that will not crash:  
1. Successful Retrieval: If the user enters a valid index, print the corresponding price and exit the program.
2. Invalid Data Type: If the user enters non-numeric text (e.g., "abc"), catch the resulting ValueError, print "Error: Please enter a valid integer index.", and prompt them again.
3. Out-of-Bounds Index: If the user enters a numeric index that falls outside the boundaries of the list (e.g., 10 or -6), catch the resulting IndexError, print "Error: Index out of range. Please choose a valid day index.", and prompt them again.<br><br>
   
The loop must continue running until a valid element is successfully retrieved from the list.

```python
stock_prices = [150.25, 153.40, 149.80, 152.10, 155.00]
while True:
    try:
        index = int(input("Enter a day index: "))
        price = stock_prices[index]
        print(price)
        break
    except ValueError:
        print("Error: Please enter a valid integer index.")
    except IndexError:
        print("Error: Index out of range. Please choose a valid day index.")
```
